```python
import numpy as np

def linear_regression(X, y):
    # Convert to numpy arrays
    X = np.array(X)
    y = np.array(y)
    
    # Add bias column (ones) to X
    n = X.shape[0]
    X_bias = np.hstack([np.ones((n, 1)), X])
    
    # Compute weights using pseudo-inverse: w = pinv(X^T X) @ X^T @ y
    XtX = X_bias.T @ X_bias
    XtX_pinv = np.linalg.pinv(XtX)
    w = XtX_pinv @ X_bias.T @ y
    
    # Round to 2 decimals and return as list
    return np.round(w, 2).tolist()
```

```python
#example 1
X = [[1], [2], [3]]
y = [2, 4, 6]
result = linear_regression(X, y)
print(result)

#example 2
X1 = [[1, 1], [2, 2], [3, 3]]
y1 = [6, 5, 12]
result1 = linear_regression(X1, y1)
print(result1)
```
